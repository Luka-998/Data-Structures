## **Linked Lists - Complete breakdown and intuition**

### 1) Singly linked list -> forward pointer from head to tail

### 2) Doubly linked list -> forward and reverse pointer from head to tail <-> tail to head

### 3) Circle linked list -> tail -> head

---

#### 1) Dynamic data structure that does not require fixed size like **arrays** that require a fixed size to be declared ahead of time. Linked list can shrink and expand during the runtime

- runtime allocation: memory is allocated for individual elements (nodes) only when they are added to the list

- No size limit

- Non-contiguous storage: Nodes are scattered at random locations in memory rather than single sequential block like arrays

- Pointer connection: Each node stores it's own data + **pointer** reference that points directly to the memory address of the next node

### **2) Run-time complextiy and basic operations implemented in the class:**

##### 2.2) delete element: O(1) time / same logic as above for array, we need to shift all elements to the left
